```python
class TrieNode:
    def __init__(self):
        self.child = [None, None]

class BitwiseTrie:
    def __init__(self):
        self.root = TrieNode()
        self.BITS = 31

    def insert(self, num):
        curr = self.root
        for i in range(self.BITS, -1, -1):
            bit = (num >> i) & 1
            if not curr.child[bit]:
                curr.child[bit] = TrieNode()

            curr = curr.child[bit]

    def search(self, num):
        curr = self.root
        for i in range(self.BITS, -1, -1):
            bit = (num >> i) & 1
            if not curr.child[bit]:
                return False
            curr = curr.child[bit]
        return True

trie = BitwiseTrie()
trie.insert(5)
trie.insert(10)
trie.insert(15)

print("Search 10: ", "Found" if trie.search(10) else "Not Found")
print("Search 7: ", "Found" if trie.search(7) else "Not Found")
```

```python

class TrieNode:
    def __init__(self):
        self.child = [None, None]


class BitwiseTrie:
    def __init__(self):
        self.root = TrieNode()
        self.BITS = 31

    def insert(self, num):
        node = self.root
        for i in range(self.BITS, -1, -1):
            bit = (num >> i) & 1
            if node.child[bit] is None:
                node.child[bit] = TrieNode()
            node = node.child[bit]

    def max_xor(self, num):
        node = self.root
        xor_sum = 0

        for i in range(self.BITS, -1, -1):
            bit = (num >> i) & 1
            opposite = 1 - bit

            if node.child[opposite]:
                xor_sum |= (1 << i)
                node = node.child[opposite]
            else:
                node = node.child[bit]

        return xor_sum


# Given array
arr = [3, 10, 5, 25]

trie = BitwiseTrie()
max_result = 0

# Insert first number
trie.insert(arr[0])

for i in range(1, len(arr)):
    max_result = max(max_result, trie.max_xor(arr[i]))
    trie.insert(arr[i])

print("Maximum XOR =", max_result)
```
