```python
import math

class MinHeap:
    def __init__(self):
        self.heap = []

    def heapify_up(self, index):
        parent = (index - 1) // 2
        if index > 0 and self.heap[parent] > self.heap[index]:
            self.heap[parent], self.heap[index] = self.heap[index], self.heap[parent]
            self.heapify_up(parent)

    def insert(self, value):
        self.heap.append(value)
        self.heapify_up(len(self.heap) - 1)
        print("\nHeap after insertion:")
        self.display_tree()

    def search(self, value):
        if value in self.heap:
            print("Element found")
        else:
            print("Element not found")

    def display_tree(self):
        n = len(self.heap)
        if n == 0:
            print("Heap is empty")
            return

        level = 0
        index = 0

        while index < n:
            count = 2 ** level
            for i in range(count):
                if index >= n:
                    break
                print(self.heap[index], end=" ")
                index += 1
            print()
            level += 1

    def display_list(self):
        print("Heap Array:", self.heap)


class PriorityQueue:
    def __init__(self):
        self.heap = MinHeap()

    def insert(self, value):
        print("\nInserting:", value)
        self.heap.insert(value)

    def search(self, value):
        self.heap.search(value)

    def show_priority_queue(self):
        sorted_list = sorted(self.heap.heap)
        print("\nPriority Queue (smallest = highest priority):")
        print(sorted_list)

    def display_tree(self):
        print("\nHeap Tree:")
        self.heap.display_tree()


pq = PriorityQueue()

while True:
    print("\n--- MENU ---")
    print("1. Insert")
    print("2. Search")
    print("3. Display Heap Tree")
    print("4. Show Priority Queue")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        val = int(input("Enter value: "))
        pq.insert(val)

    elif choice == 2:
        val = int(input("Enter value to search: "))
        pq.search(val)

    elif choice == 3:
        pq.display_tree()

    elif choice == 4:
        pq.show_priority_queue()

    elif choice == 5:
        print("Exiting...")
        break

    else:
        print("Invalid choice!")
```
