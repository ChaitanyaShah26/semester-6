```python
class CuckooHash:
    def __init__(self, size):
        self.size = size
        self.table1 = [None] * size
        self.table2 = [None] * size
        self.max_attempts = size * 2 # Increase max_attempts for better chance to insert

    def h1(self, key):
        return key % self.size

    def h2(self, key):
        return (key // self.size) % self.size

    def insert(self, key):
        original_key = key
        for attempt in range(self.max_attempts):
            # Try inserting into table1
            pos1 = self.h1(key)
            if self.table1[pos1] is None:
                self.table1[pos1] = key
                print(f"Key {original_key} inserted successfully.")
                return True

            # Evict existing key from table1
            evicted_key = self.table1[pos1]
            self.table1[pos1] = key
            key = evicted_key # The evicted key now needs to be re-inserted

            # Try inserting the evicted key into table2
            pos2 = self.h2(key)
            if self.table2[pos2] is None:
                self.table2[pos2] = key
                print(f"Key {original_key} inserted successfully (after evicting {evicted_key}).")
                return True

            # Evict existing key from table2
            evicted_key = self.table2[pos2]
            self.table2[pos2] = key
            key = evicted_key # The newly evicted key now needs to be re-inserted

        print(f"Cycle detected! Failed to insert key {original_key} after multiple attempts. Rehash required.")
        return False

    def search(self, key):
        if self.table1[self.h1(key)] == key:
            return True
        if self.table2[self.h2(key)] == key:
            return True
        return False

    def delete(self, key):
        pos1 = self.h1(key)
        if self.table1[pos1] == key:
            self.table1[pos1] = None
            print(f"Key {key} deleted from Table 1.")
            return True
        pos2 = self.h2(key)
        if self.table2[pos2] == key:
            self.table2[pos2] = None
            print(f"Key {key} deleted from Table 2.")
            return True
        print(f"Key {key} not found for deletion.")
        return False

    def display(self):
        print("\n--- Cuckoo Hash Table State ---")
        print("Table 1:", self.table1)
        print("Table 2:", self.table2)
        print("-------------------------------")

# Interactive usage
hash_table = CuckooHash(size=7)

while True:
    print("\nCuckoo Hashing Menu:")
    print("1. Insert key")
    print("2. Search key")
    print("3. Delete key")
    print("4. Display tables")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        try:
            key = int(input("Enter key to insert: "))
            hash_table.insert(key)
            hash_table.display()
        except ValueError:
            print("Invalid input. Please enter an integer.")
    elif choice == '2':
        try:
            key = int(input("Enter key to search: "))
            if hash_table.search(key):
                print(f"Key {key} found in the hash table.")
            else:
                print(f"Key {key} not found in the hash table.")
            hash_table.display() # Display after search to show current state
        except ValueError:
            print("Invalid input. Please enter an integer.")
    elif choice == '3':
        try:
            key = int(input("Enter key to delete: "))
            hash_table.delete(key)
            hash_table.display()
        except ValueError:
            print("Invalid input. Please enter an integer.")
    elif choice == '4':
        hash_table.display()
    elif choice == '5':
        print("Exiting Cuckoo Hashing program.")
        break
    else:
        print("Invalid choice. Please try again.")
```
