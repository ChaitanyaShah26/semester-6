```python
cap = 3
size = 0
cost = 0
arr = [0] * cap
print("Inserting elements: ")
for i in range(14):
    if size == cap:
        cap *= 2
        new_arr = [0] * cap
        for j in range(size):
            new_arr[j] = arr[j]
            cost += 1
        arr = new_arr
        print(f"Resized array to capacity : {cap}")
    arr[size] = i
    size += 1
    cost += 1
    print(f"Inserted {i} : Cost {cost}")
```
