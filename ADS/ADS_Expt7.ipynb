```python
import collections

class KDNode:
    def __init__(self, point, axis, left=None, right=None):
        self.point = point
        self.axis = axis
        self.left = left
        self.right = right

    def __repr__(self):
        return f"KDNode(point={self.point}, axis={self.axis})"

class KDTree:
    def __init__(self, dimensions):
        self.root = None
        self.dimensions = dimensions

    def _insert_recursive(self, node, point, depth):
        if node is None:
            return KDNode(point, depth % self.dimensions)

        current_axis = depth % self.dimensions
        if point[current_axis] < node.point[current_axis]:
            node.left = self._insert_recursive(node.left, point, depth + 1)
        else:
            node.right = self._insert_recursive(node.right, point, depth + 1)
        return node

    def insert(self, point):
        if not isinstance(point, (list, tuple)) or len(point) != self.dimensions:
            raise ValueError(f"Point must be a list/tuple of {self.dimensions} dimensions.")
        self.root = self._insert_recursive(self.root, point, 0)
        print(f"Point {point} inserted.")
        self.display_tree()

    def _search_recursive(self, node, point, depth):
        if node is None:
            return False
        if node.point == point:
            return True

        current_axis = depth % self.dimensions
        if point[current_axis] < node.point[current_axis]:
            return self._search_recursive(node.left, point, depth + 1)
        else:
            return self._search_recursive(node.right, point, depth + 1)

    def search(self, point):
        if not isinstance(point, (list, tuple)) or len(point) != self.dimensions:
            raise ValueError(f"Point must be a list/tuple of {self.dimensions} dimensions.")
        found = self._search_recursive(self.root, point, 0)
        if found:
            print(f"Point {point} found in the KD Tree.")
        else:
            print(f"Point {point} not found in the KD Tree.")
        return found

    def _find_min_recursive(self, node, target_axis, current_axis, depth):
        if node is None:
            return None

        if current_axis == target_axis:
            if node.left is None:
                return node.point
            else:
                return self._find_min_recursive(node.left, target_axis, (depth + 1) % self.dimensions, depth + 1)
        else:
            min_point = node.point
            left_min = self._find_min_recursive(node.left, target_axis, (depth + 1) % self.dimensions, depth + 1)
            right_min = self._find_min_recursive(node.right, target_axis, (depth + 1) % self.dimensions, depth + 1)

            if left_min and left_min[target_axis] < min_point[target_axis]:
                min_point = left_min
            if right_min and right_min[target_axis] < min_point[target_axis]:
                min_point = right_min
            return min_point

    def find_min(self, target_axis):
        if not (0 <= target_axis < self.dimensions):
            raise ValueError(f"Target axis must be between 0 and {self.dimensions - 1}.")
        if self.root is None:
            print("Tree is empty. No minimum element.")
            return None

        min_val = self._find_min_recursive(self.root, target_axis, 0, 0)
        print(f"Minimum point along axis {target_axis} is: {min_val}")
        return min_val

    def _print_tree_recursive(self, node, level, prefix):
        if node is not None:
            print('    ' * level + prefix + f"P:{node.point} A:{node.axis}")
            if node.left or node.right:
                self._print_tree_recursive(node.left, level + 1, 'L-- ')
                self._print_tree_recursive(node.right, level + 1, 'R-- ')

    def display_tree(self):
        if self.root is None:
            print("KD Tree is empty.")
            return

        print("\n--- KD Tree Structure (Hierarchical) ---")
        self._print_tree_recursive(self.root, 0, 'Root: ')
        print("----------------------------------------")


def get_point_input(dimensions):
    while True:
        try:
            point_str = input(f"Enter {dimensions} comma-separated numerical values for the point: ")
            point = [float(x.strip()) for x in point_str.split(',')]
            if len(point) != dimensions:
                print(f"Error: Please enter exactly {dimensions} values.")
                continue
            return tuple(point)
        except ValueError:
            print("Invalid input. Please enter numerical values separated by commas.")

def get_dimension_input():
    while True:
        try:
            dim = int(input("Enter the number of dimensions for the KD Tree (e.g., 2 for 2D, 3 for 3D): "))
            if dim <= 0:
                print("Dimensions must be a positive integer.")
                continue
            return dim
        except ValueError:
            print("Invalid input. Please enter an integer.")

def get_axis_input(dimensions):
    while True:
        try:
            axis = int(input(f"Enter the axis (0 to {dimensions - 1}) to find the minimum: "))
            if 0 <= axis < dimensions:
                return axis
            else:
                print(f"Invalid axis. Please enter a value between 0 and {dimensions - 1}.")
        except ValueError:
            print("Invalid input. Please enter an integer.")

def main():
    print("Welcome to the KD Tree Implementation!")
    dimensions = get_dimension_input()
    kdtree = KDTree(dimensions)

    while True:
        print("\nMenu:")
        print("1. Insert a point")
        print("2. Search for a point")
        print("3. Find minimum along an axis")
        print("4. Display Tree (current state)")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            point = get_point_input(dimensions)
            kdtree.insert(point)
        elif choice == '2':
            point = get_point_input(dimensions)
            kdtree.search(point)
        elif choice == '3':
            axis = get_axis_input(dimensions)
            kdtree.find_min(axis)
        elif choice == '4':
            kdtree.display_tree()
        elif choice == '5':
            print("Exiting KD Tree program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == '__main__':
    main()
```
